```python
from scripts_.stitching_vk_data import plot_stitched_height_image_from_vk #, plot_stitched_height_image_from_vk_2
from scripts_.tilt_correction import baseplane_filter
from scripts_.plot_vk_data import plot_vk4_height, plot_vk4_reflected_light, reshape_img_data
from scripts_.plot_csv_data import plot_csv_height
from scripts_.load_vk_data import parse_vk6_file, parse_vk4_file
from scripts_.preprocess_vk_data import crop_Vk4File_map
from scripts_.feature_enhancement import feature_enhancement_proceedings, feature_enhancement_1#, feature_enhancement_2, feature_enhancement_3
from scripts_.vk_height_to_csv import vk_height_to_csv

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter
```

```python
folder = Path("../all_raw_data_vk/Muscovite mica/Stitched 50x region 3")

for image_path in folder.iterdir():
    if image_path.is_file():
        #file_name = str(image_path.name)
        with open(image_path, 'rb') as f:
            data_vk4 = parse_vk4_file(f)

        data_vk4_tiltcorr = baseplane_filter(data_vk4, 90)
        data_vk4_enhanced = feature_enhancement_proceedings(data_vk4_tiltcorr)

    ax = plot_vk4_height(data_vk4_tiltcorr, data_type='baseplane_subtracted')
    plt.show()

    reshaped = reshape_img_data(data_vk4)
    profile = np.mean(reshaped, axis=0) #nanmedian
    x_pixels = reshaped.shape[1]
    
    percentages = [10, 20, 40, 80, 100]

    for percentage in percentages:
        data_1_vk4_tiltcorr = baseplane_filter(data_vk4, percentage)
        reshaped = reshape_img_data(data_1_vk4_tiltcorr, data_type='baseplane_fit')
        profile2 = np.mean(reshaped, axis=0) #nanmedian
        x_pixels = reshaped.shape[1]    
        plt.plot(range(x_pixels), profile2, label=f'{percentage}%')
    plt.plot(range(x_pixels), profile, color='lightgreen', label='raw data')
    plt.title('Profile of baseplane fit + raw data')
    plt.xlabel('X (pixels)')
    plt.ylabel('Height (µm)')
    plt.legend()
    plt.show()
```
